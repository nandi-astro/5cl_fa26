```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
d0_array = np.array([36, 38.5, 33.5, 29.5, 49, 23])
di_array = np.array([38.5,26,28,32,22,48])
hi_array = np.array([2.2, 1.1, 1, 1.4, 0.8, 3.2])
h0_array = np.array([0.8, 0.8, 0.8, 0.8, 0.8, 0.8])
```

```python
x = 1/d0_array
y = 1/di_array
```

```python
#determine covariance and coefficient of linear correlation
cov_xy = np.cov(x,y)[0]
rho_xy = np.corrcoef(x,y)[0][1]
print("sigma_xy: \n",cov_xy)
print("rho_xy: \n",rho_xy)
```

```python
alpha_xi = 0.5/(d0_array**2)
alpha_yi = 0.2/(di_array**2)
alpha_y_correct = np.sqrt(alpha_yi**2 + alpha_xi**2)
print(alpha_y_correct)
```

```python
wi = 1/(alpha_y_correct**2)
b = (np.sum(wi*(y+x)))/np.sum(wi)
alpha_b = 1/np.sqrt(np.sum(wi))
print(b)
print(alpha_b)
```

```python
y_model = -1*x + b
chi2 = np.sum(((y - y_model) / alpha_y_correct)**2)
dof = len(y) - 1
chi2_reduced = chi2 / dof
print(chi2)
print(dof)
print(chi2_reduced)
```

```python
focal = 1 / b
alpha_f = alpha_b / (b**2)
print(focal)
print(alpha_f)
```
